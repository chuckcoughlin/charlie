```python
from boosteros.robots.booster import BoosterRobot
from boosteros.types import ImageType

# or robot = K1Robot()
robot = BoosterRobot()

image = robot.get_image()
image_depth = robot.get_image(img_type=ImageType.DEPTH)
```

```python
from boosteros.brain import Detection

print("开始图像处理")
detection = Detection(model="soccer", backend="local")
boxes = detection.detect(image.to_numpy())
print("boxes", boxes)
detection.plot(image.to_numpy(), boxes)
```

```python
import time
from datetime import datetime


def move_by_distance(dist: int, speed: float, step: float = 0.5):
    cur_mode = robot.get_mode()
    print("current mode", cur_mode)
    if cur_mode != "walk":
        robot.set_mode("walk")

    robot.reset_odom()
    print("reset odom")

    while True:
        odom = robot.get_odom()
        print(
            f"{datetime.now().strftime('%Y-%m-%d %H:%M:%S.%f')[:-3]} pose_2d({odom.pose_2d[0]:.3f}, {odom.pose_2d[1]:.3f}, {odom.pose_2d[2]:.3f})"
        )

        if odom.pose_2d[0] > dist:
            robot.set_velocity(0.0, 0.0, 0.0)
            print("到达目标位置，停止运动")
            break

        robot.set_velocity(speed, 0, 0)
        time.sleep(step)

if __name__ == "__main__":
    move_by_distance(2, 0.1, 0.1)
```
